# **Import Pandas**

```python
import pandas as pd
```

# **Load the CSV file**

```python
df = pd.read_csv('/content/BCG.csv')

df
```

# **Analyzing trends with pandas**

```python
# Financial columns that should contain numbers
metrics = [
    'Total Revenue',
    'Net Income',
    'Total Assets',
    'Total Liabilities',
    'Operating Cash Flow'
]

# Remove commas and convert financial columns to numbers
for metric in metrics:
    df[metric] = (
        df[metric]
        .astype(str)
        .str.replace(',', '', regex=False)
        .str.strip()
    )
    df[metric] = pd.to_numeric(df[metric], errors='coerce')

# Make sure Fiscal Year is numeric
df['Fiscal Year'] = pd.to_numeric(df['Fiscal Year'], errors='coerce')
```

# **Check the data**

```python
print("Dataset shape:")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nFirst five rows:")
display(df.head())
```

# **Sort the data**

```python
df = df.sort_values(
    ['Company', 'Fiscal Year']
).reset_index(drop=True)
```

# **Calculate Year-over-Year Growth**

```python
df['Revenue Growth (%)'] = (
    df.groupby('Company')['Total Revenue']
    .pct_change() * 100
)

df['Net Income Growth (%)'] = (
    df.groupby('Company')['Net Income']
    .pct_change() * 100
)

df['Assets Growth (%)'] = (
    df.groupby('Company')['Total Assets']
    .pct_change() * 100
)

df['Liabilities Growth (%)'] = (
    df.groupby('Company')['Total Liabilities']
    .pct_change() * 100
)

df['Operating Cash Flow Growth (%)'] = (
    df.groupby('Company')['Operating Cash Flow']
    .pct_change() * 100
)
```

# **Round percentage values**

```python
# Find the highest and lowest growth for each financial metric

growth_columns = [
    'Revenue Growth (%)',
    'Net Income Growth (%)',
    'Assets Growth (%)',
    'Liabilities Growth (%)',
    'Operating Cash Flow Growth (%)'
]

for column in growth_columns:
    print("\n", column)
    print("Highest growth:")
    display(df.loc[df[column].idxmax(), ['Company', 'Fiscal Year', column]])

    print("Lowest growth:")
    display(df.loc[df[column].idxmin(), ['Company', 'Fiscal Year', column]])
```

```python
print("\nComplete Financial Dataset with Growth Rates:")
display(df)
```

# **Company-Level Average Analysis**

```python
company_summary = df.groupby('Company')[
    [
        'Total Revenue',
        'Net Income',
        'Total Assets',
        'Total Liabilities',
        'Operating Cash Flow'
    ]
].mean().round(2)

print("\nAverage Financial Performance by Company:")
display(company_summary)
```

# **2023 to 2025 Percentage Change**

```python
start = df[df['Fiscal Year'] == 2023].set_index('Company')
end = df[df['Fiscal Year'] == 2025].set_index('Company')

change = pd.DataFrame(index=start.index)

for metric in metrics:
    change[f'{metric} Change (%)'] = (
        (end[metric] / start[metric]) - 1
    ) * 100

change = change.round(2)

print("\nPercentage Change from 2023 to 2025:")
display(change)
```

# **Financial Performance by Year**

```python
year_summary = df.groupby('Fiscal Year')[
    [
        'Total Revenue',
        'Net Income',
        'Total Assets',
        'Total Liabilities',
        'Operating Cash Flow'
    ]
].sum().round(2)

print("\nCombined Financial Performance by Year:")
display(year_summary)

```
